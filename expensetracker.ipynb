```python
import tkinter as tk
from tkinter import ttk, messagebox
import csv
import os
from datetime import date

FILE = "expenses.csv"

# Main window
root = tk.Tk()
root.title("Expense Tracker")
root.geometry("650x700")
root.configure(bg="#EAF2F8")
root.resizable(False, False)

expenses = []

# Save expenses
def save_expenses():
    with open(FILE, "w", newline="") as f:
        writer = csv.writer(f)
        writer.writerow(["Date", "Description", "Category", "Amount"])
        writer.writerows(expenses)

# Load expenses
def load_expenses():
    if os.path.exists(FILE):
        with open(FILE, "r", newline="") as f:
            reader = csv.reader(f)
            next(reader, None)
            for row in reader:
                if len(row) == 4:
                    expenses.append(row)
    refresh_list()

# Update display
def refresh_list():
    expense_list.delete(0, tk.END)
    total = 0

    for item in expenses:
        expense_list.insert(
            tk.END,
            f"{item[0]} | {item[1]} | {item[2]} | Rs. {item[3]}"
        )
        total += float(item[3])

    total_label.config(
        text=f"Total Expenses: Rs. {total:.2f}"
    )

# Add expense
def add_expense():
    description = desc_entry.get().strip()
    amount = amount_entry.get().strip()
    category = category_box.get()

    try:
        value = float(amount)

        if not description or not category:
            messagebox.showwarning(
                "Warning", "Please fill all fields!"
            )
            return

        if value <= 0:
            messagebox.showwarning(
                "Warning", "Enter an amount greater than zero!"
            )
            return

        expenses.append([
            str(date.today()),
            description,
            category,
            f"{value:.2f}"
        ])

        save_expenses()
        refresh_list()

        desc_entry.delete(0, tk.END)
        amount_entry.delete(0, tk.END)

    except ValueError:
        messagebox.showerror(
            "Invalid Input",
            "Please enter a valid amount!"
        )

# Delete expense
def delete_expense():
    selected = expense_list.curselection()

    if selected:
        expenses.pop(selected[0])
        save_expenses()
        refresh_list()
    else:
        messagebox.showwarning(
            "Warning", "Please select an expense!"
        )

# Heading
title = tk.Label(
    root,
    text="EXPENSE TRACKER",
    font=("Arial", 23, "bold"),
    bg="#1F4E79",
    fg="white",
    pady=18
)
title.pack(fill="x")

# Main frame
frame = tk.Frame(root, bg="white", padx=25, pady=20)
frame.pack(padx=25, pady=20, fill="both", expand=True)

# Description
tk.Label(
    frame, text="Expense Description",
    bg="white", font=("Arial", 11, "bold")
).pack(anchor="w")

desc_entry = tk.Entry(frame, font=("Arial", 12))
desc_entry.pack(fill="x", pady=5)

# Amount
tk.Label(
    frame, text="Amount (Rs.)",
    bg="white", font=("Arial", 11, "bold")
).pack(anchor="w")

amount_entry = tk.Entry(frame, font=("Arial", 12))
amount_entry.pack(fill="x", pady=5)

# Category
tk.Label(
    frame, text="Select Category",
    bg="white", font=("Arial", 11, "bold")
).pack(anchor="w")

category_box = ttk.Combobox(
    frame,
    values=[
        "Food", "Travel", "Shopping",
        "Education", "Bills", "Other"
    ],
    state="readonly",
    font=("Arial", 11)
)
category_box.pack(fill="x", pady=5)
category_box.set("Food")

# Add button
tk.Button(
    frame,
    text="ADD EXPENSE",
    command=add_expense,
    bg="#2980B9",
    fg="white",
    font=("Arial", 11, "bold"),
    pady=7
).pack(fill="x", pady=10)

# Expense list
tk.Label(
    frame, text="Expense History",
    bg="white", font=("Arial", 12, "bold")
).pack(anchor="w")

expense_list = tk.Listbox(
    frame,
    font=("Arial", 10),
    height=5,
    width=65
)
expense_list.pack(fill="x", pady=8)

# Delete button
tk.Button(
    frame,
    text="DELETE SELECTED",
    command=delete_expense,
    bg="#E67E22",
    fg="white",
    font=("Arial", 10, "bold"),
    pady=6
).pack(fill="x", pady=5)

# Total expenses
total_label = tk.Label(
    frame,
    text="Total Expenses: Rs. 0.00",
    font=("Arial", 15, "bold"),
    bg="white",
    fg="#16803C"
)
total_label.pack(pady=12)

# Footer
tk.Label(
    root,
    text="Developed using Python Tkinter",
    bg="#EAF2F8",
    fg="#555555"
).pack(pady=5)

# Load saved data
load_expenses()

# Run application
root.mainloop()
```
